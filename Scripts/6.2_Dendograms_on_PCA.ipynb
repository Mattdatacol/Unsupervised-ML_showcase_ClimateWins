## 1. Libraries and Data importations
<div id='1'></div>

```python
import pandas as pd
import numpy as np
import seaborn as sns
import os
import operator
import matplotlib.pyplot as plt
from sklearn import datasets
from sklearn import metrics
from sklearn.preprocessing import StandardScaler
from scipy.cluster.hierarchy import dendrogram, linkage, fcluster
from sklearn.metrics import silhouette_score


##to ignore warnings
import warnings
warnings.filterwarnings('ignore')

# Creating paths to the data folder of the project and to the folder for saving the charts
path_data = r"C:\Users\dacol\Documents\Data_Project_careerfoundry\ClimateWins_ML\02 Data"

# Importing the PCA scaled file
dfPCA = pd.read_pickle(os.path.join(path_data,'Prepared Data','PCA_90s.pkl'))
```

## 2. Running the 4 dendograms methods
<div id='2'></div>

### 2.1 Dendograms on all metrics
<div id='2.1'></div>

```python
methods = ['single', 'complete', 'average', 'ward']
for method in methods:
    plt.figure(figsize=(18, 6))
    distance = linkage(dfPCA, method=method)
    dendrogram(distance, leaf_rotation=90)
    plt.xlabel('Index')
    plt.ylabel('Distance')
    plt.title(f'Dendrogram {method.capitalize()} Method')
    plt.show()
    
    # Create clusters
    clusters = fcluster(distance, 2, criterion='maxclust')
    
    # Calculate silhouette score
    silhouette_avg = silhouette_score(dfPCA, clusters)
    print(f'Silhouette Score for {method.capitalize()} method: {silhouette_avg:.3f}')
```

### 2.2 Dendogram on each PCA
<div id='2.2'></div>

```python
# Create a list of column names
pca_columns = dfPCA.columns
pca_columns
```

```python
for pca in pca_columns:
    
    # Create dendrograms and clusters for each method
    for method in methods:
        distance = linkage(dfPCA, method=method)
        
        # Plot dendrogram
        plt.figure(figsize=(18, 6))
        dendrogram(distance, leaf_rotation=90)
        plt.xlabel('Index')
        plt.ylabel('Distance')
        plt.title(f'Dendrogram {method.capitalize()} Method - {pca}')
        plt.show()
        
        # Create clusters
        dfPCA[f'cluster_{method}'] = fcluster(distance, 2, criterion='maxclust')

        # Compute silhouette score
        silhouette_avg = silhouette_score(
        dfPCA,
        dfPCA[f'cluster_{method}']
        )
        print(f'Cluster Distribution - {method.capitalize()} Method\nSilhouette Score: {silhouette_avg:.3f}')
```
